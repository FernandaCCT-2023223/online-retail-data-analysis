## Identify Invalid values


```python
#identify record with negative or zero quantities
invalid_quantities = df[df["Quantity"] <= 0]

#count invalid quantity records
invalid_quantities.shape[0]
```

```python
# Identify records with negative or zero unit prices
invalid_prices = df[df["UnitPrice"] <= 0]

#Count invalid price records 
invalid_prices.shape[0]
```

```python
# Count negative and zero quantities
negative_quantities = (df["Quantity"] < 0).sum()
zero_quantities = (df["Quantity"] == 0).sum()

# Display the results
print("Negative quantities: ", negative_quantities)
print("Zero quantities: ", zero_quantities)
```

```python
#Count the negatives unit pricess
negative_prices = (df["UnitPrice"] < 0).sum()
zero_prices = (df["UnitPrice"] == 0).sum()

# Display the results
print("Negative prices:", negative_prices)
print("Z ero prices:", zero_prices)
```
