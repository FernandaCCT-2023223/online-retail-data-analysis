# Data Cleaning abd Data Transformation


```python
# Create a copy of the original dataset
df_clean = df.copy()

# Check the initial number of records 
print("Original records: ", len(df))
print("Working dataset records: ", len(df_clean))
```

```python
# Remove duplicate records 
df_clean = df_clean.drop_duplicates()

#chect the number of records after removing duplicates
print("Records after removing duplicates: ", len(df_clean))

#check remaning duplicate records
print("Remaining duplicates: ", df_clean.duplicated().sum())
```

### Missing values Treatment

```python
# Check missing values in the cleaned dataset
missing_values_clean = df_clean.isnull().sum()

# Display missing values
missing_values_clean
```

```python
# Filter records with missing descriptions
missing_descriptions = df_clean[df_clean["Description"].isnull()]

#Display the first 10 records
missing_descriptions.head(10)
```

```python
# Count unique stock codes with missing descriptions
missing_descriptions["StockCode"].nunique()
```

### Check description recovery

```python
# Identify stock codes with only one known description
description_lookup = (
    df_clean.dropna(subset=["Description"])
    .groupby("StockCode")["Description"]
    .nunique()
)

# Select stock codes with a unique description
recoverable_stock_codes = description_lookup[
    description_lookup == 1
].index

# Count missing descriptions that can potentially be recovered
recoverable_records = missing_descriptions[
    missing_descriptions["StockCode"].isin(recoverable_stock_codes)
]

print("Recoverable records:", len(recoverable_records))
print("Recoverable stock codes:", recoverable_records["StockCode"].nunique())
```

### Recover missing descriptions 

```python
# Create a lookup table for recoverable product descriptions
description_map = (
    df_clean[
        df_clean["StockCode"].isin(recoverable_stock_codes)
    ]
    .dropna(subset=["Description"])
    .drop_duplicates(subset=["StockCode"])
    .set_index("StockCode")["Description"]
)

# Fill missing descriptions using the StockCode
df_clean["Description"] = df_clean["Description"].fillna(
    df_clean["StockCode"].map(description_map)
)

# Check the remaining missing descriptions
print("Missing descriptions after recovery:",
      df_clean["Description"].isnull().sum())
```

### Investigate remaining missing descriptions

```python
# Filter records with missing product descriptions
missing_descriptions = df_clean[
    df_clean["Description"].isnull()
]

# Display the number of affected records
print("Records with missing descriptions:", len(missing_descriptions))

# Count unique stock codes
print("Unique stock codes:",
      missing_descriptions["StockCode"].nunique())

# Display the first 15 records
missing_descriptions.head(15)
```

```python
# Analyse the remaining records with missing descriptions

print("Zero unit prices:",
      (missing_descriptions["UnitPrice"] == 0).sum())

print("Negative unit prices:",
      (missing_descriptions["UnitPrice"] < 0).sum())

print("Negative quantities:",
      (missing_descriptions["Quantity"] < 0).sum())

print("Positive quantities:",
      (missing_descriptions["Quantity"] > 0).sum())
```
