# Data Cleaning abd Data Transformation


```python
# Create a copy of the original dataset
df_clean = df.copy()

# Check the initial number of records 
print("Original records: ", len(df))
print("Working dataset records: ", len(df_clean))
```

```python
# Remove duplicate records 
df_clean = df_clean.drop_duplicates()

#chect the number of records after removing duplicates
print("Records after removing duplicates: ", len(df_clean))

#check remaning duplicate records
print("Remaining duplicates: ", df_clean.duplicated().sum())
```

### Missing values Treatment

```python
# Check missing values in the cleaned dataset
missing_values_clean = df_clean.isnull().sum()

# Display missing values
missing_values_clean
```

```python
# Filter records with missing descriptions
missing_descriptions = df_clean[df_clean["Description"].isnull()]

#Display the first 10 records
missing_descriptions.head(10)
```

```python
# Count unique stock codes with missing descriptions
missing_descriptions["StockCode"].nunique()
```

### Check description recovery

```python
# Identify stock codes with only one known description
description_lookup = (
    df_clean.dropna(subset=["Description"])
    .groupby("StockCode")["Description"]
    .nunique()
)

# Select stock codes with a unique description
recoverable_stock_codes = description_lookup[
    description_lookup == 1
].index

# Count missing descriptions that can potentially be recovered
recoverable_records = missing_descriptions[
    missing_descriptions["StockCode"].isin(recoverable_stock_codes)
]

print("Recoverable records:", len(recoverable_records))
print("Recoverable stock codes:", recoverable_records["StockCode"].nunique())
```

### Recover missing descriptions 

```python
# Create a lookup table for recoverable product descriptions
description_map = (
    df_clean[
        df_clean["StockCode"].isin(recoverable_stock_codes)
    ]
    .dropna(subset=["Description"])
    .drop_duplicates(subset=["StockCode"])
    .set_index("StockCode")["Description"]
)

# Fill missing descriptions using the StockCode
df_clean["Description"] = df_clean["Description"].fillna(
    df_clean["StockCode"].map(description_map)
)

# Check the remaining missing descriptions
print("Missing descriptions after recovery:",
      df_clean["Description"].isnull().sum())
```

### Missing CustomerID

During the data inspection, I found that 135,037 records (25.16%) have a missing CustomerID.

I decided to keep these records because they still contain useful information about the transactions, such as the products purchased, quantities, prices, dates and countries. Removing them would mean losing a significant amount of sales data.

But, these records cannot be used when analysing individual customers, as we do not know which customer made the purchase.

```python
# Identify invalid transactions

cancelled_mask = (
    df_clean["InvoiceNo"].astype(str).str.startswith("C")
)

invalid_quantity_mask = df_clean["Quantity"] <= 0

invalid_price_mask = df_clean["UnitPrice"] <= 0

# Count records in each category

print("Cancelled invoices:", cancelled_mask.sum())
print("Invalid quantities:", invalid_quantity_mask.sum())
print("Invalid prices:", invalid_price_mask.sum())
```

### Identify unique ivalid records

```python
# Combine all invalid transaction conditions

invalid_transactions_mask = (
    cancelled_mask |
    invalid_quantity_mask |
    invalid_price_mask
)

# Count unique invalid records

print("Total invalid records:",
      invalid_transactions_mask.sum())

print("Records remaining:",
      len(df_clean) - invalid_transactions_mask.sum())
```

### Removing invalid transactions

After checking the cancelled invoices, quantities and unit prices, I identified 11,763 invalid records. These include cancelled transactions, negative or zero quantities, and negative or zero unit prices.

I decided to remove these records because they do not represent valid sales for the revenue analysis. This will help make the results more reliable.

```python
# remove invalid transactions
df_clean = df_clean[~invalid_transactions_mask].copy()

#check the number of records after cleaning
print("Records after removing invalid transactions:", len(df_clean))

# Check if any invalid transactions remain
print("Remaining invalid transactions", 
      (
          df_clean["InvoiceNo"].astype(str).str.startswith("C") |
          (df_clean["Quantity"] <= 0) |
          (df_clean["UnitPrice"] <= 0)
      ).sum())
          
```

## Outlier Detection

```python
Q1 = df_clean["Quantity"].quantile(0.25)
Q3 = df_clean["Quantity"].quantile(0.75)

IQR = Q3 - Q1

lower_limit = Q1 - 1.5 * IQR
upper_limit = Q3 + 1.5 * IQR

quantity_outliers = df_clean[
    (df_clean["Quantity"] < lower_limit) |
    (df_clean["Quantity"] > upper_limit)
]

print("Q1:", Q1)
print("Q3:", Q3)
print("IQR:", IQR)
print("Lower limit:", lower_limit)
print("Upper limit:", upper_limit)
print("Number of outliers:", len(quantity_outliers))
```

```python
Q1_price = df_clean["UnitPrice"].quantile(0.25)
Q3_price = df_clean["UnitPrice"].quantile(0.75)

IQR_price = Q3_price - Q1_price

lower_price = Q1_price - 1.5 * IQR_price
upper_price = Q3_price + 1.5 * IQR_price

price_outliers = df_clean[
    (df_clean["UnitPrice"] < lower_price) |
    (df_clean["UnitPrice"] > upper_price)
]

print("Q1:", Q1_price)
print("Q3:", Q3_price)
print("IQR:", IQR_price)
print("Lower limit:", lower_price)
print("Upper limit:", upper_price)
print("Number of price outliers:", len(price_outliers))
```
